```python
from pathlib import Path
import os
import pandas as pd

try:
    PROJECT_DIR = Path(__file__).resolve().parents[0]
except NameError:
    PROJECT_DIR = Path(os.getcwd()).resolve().parents[0]
DATA_PATH = PROJECT_DIR / "data" / "online_retail.csv"

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATA_PATH}\n"
        "Place Online Retail.xlsx inside the data folder."
    )
else: print(f"Dataset in {DATA_PATH}")
```

```python
df = pd.read_csv(DATA_PATH)
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(f"\n------------Column names-------------\n{df.columns.tolist()}")
print(f"\n------------First 5 records----------\n{df.head().to_string()}")
print(f"\n------------Data types---------------\n{df.dtypes}")
print(f"-----------Missing Values--------------\n{df.isna().sum()}")
print(f"Duplicated Rows: {df.duplicated().sum():,}")
print(f"Negative Quantities: {(df["Quantity"] < 0).sum()}")
print(f"Zero or negative prices: {(df["UnitPrice"] <= 0).sum()}")
print(f"----------------Numerical summary-----------\n{df[["Quantity", "UnitPrice"]].describe()}")
print(f"\n-----------------------Invoice checks-----------")
print(f"Cancelled invoices: {df["InvoiceNo"].astype(str).str.startswith("C").sum()}")
print(f"----------Date Range----------\nEarliest: {df["InvoiceDate"].min()}\nLatest: {df['InvoiceDate'].max()}")

```
